```python
import pandas as pd, sqlite3
```

```python
conn = sqlite3.connect("../data/f1.db")
```

```python
tables = ["races", "results", "drivers", "constructors", "circuits", "status", "qualifying", "pit_stops"]
```

```python
for t in tables:
    df = pd.read_csv(f"../data/raw/{t}.csv", na_values="\\N")
    df.to_sql(t, conn, if_exists="replace", index=False)
    print(t, df.shape)
conn.close()
```
